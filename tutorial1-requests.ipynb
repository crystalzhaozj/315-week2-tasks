### 2. Inspect the Response

Response = object that contains the results of your request

```python
response = requests.get("https://api.github.com")
```

Status code: 200 = successful; 404 = not found; 204 = successful response but no contet; 304 = not modified.  

Could also be used in a Boolean context, which is the default behavior of Response.

Raising an error if unsure what's going wrong:

```python
from requests.exceptions import HTTPError

URLS = ["https://api.github.com", "https://api.github.com/invalid"]

for url in URLS:
    try:
        response = requests.get(url)
        response.raise_for_status()
    except HTTPError as http_err:
        print(f"HTTP error occurred: {http_err}")
    except Exception as err:
        print(f"Other error occurred: {err}")
    else:
        print("Success!")
```

### 4. Add Query String Parameters

How do you want your response to long? 

```python
response = requests.get(
    "https://api.github.com/search/repositories",
    params={"q": "language:python", "sort": "stars", "order": "desc"},
)

json_response = response.json()
popular_repositories = json_response["items"]
for repo in popular_repositories[:3]:
    print(f"Name: {repo['name']}")
    print(f"Description: {repo['description']}")
    print(f"Stars: {repo['stargazers_count']}\n")
```

Can pass as dictionary, tuples, or even bytes.

### 5. Customize Request Headers

Request header is an HTTP header that can be used in an HTTP request to provide information about the request context, so the server can tailor the response. 

Accept header tells the server what content types your application can handle.

```python
response = requests.get(
    "https://api.github.com/search/repositories",
    params={"q": '"real python"'},
    headers={"Accept": "application/vnd.github.text-match+json"},
)

json_response = response.json()
first_repository = json_response["items"][0]
print(first_repository["text_matches"][0]["matches"])
```

### Improve Performance

Pay attention to how timeout control, sessions, and retry limits can help keep the application running smoothly.

**Setting request timeouts:**

```python
requests.get("https://api.github.com", timeout=1)
```

This works because the server responded within the 1 second timeout limit.

```python
requests.get("https://api.github.com", timeout=0.01)
```

This does NOT work because the server cannot complete it so quickly.

**Reuse Connections with Session Objects**

What are sessions? To fine-tune your control over hw requests are made or to improve the performance of requests, we need to use a Session instance directly (a request is an abstraction of Session).

You can keep the parameters persistent across requests.

**Retry Failed Requests**

Implement a custom transport adapter: define a set of configurations for each service you interact with.

First define your retry strategy (how many times, which statuses would trigger retries..);

The instantiate an adapter which would be mounted to session, so the session adheres to configuration for each request to the URL.

```python
from requests.adapters import HTTPAdapter
from requests.exceptions import RetryError
from urllib3.util.retry import Retry

retry_strategy = Retry(
    total=2, # try a maximum number of two times
    status_forcelist=[429, 500, 502, 503, 504]
)
github_adapter = HTTPAdapter(max_retries=retry_strategy)

with requests.Session() as session:
    session.mount("https://api.github.com", github_adapter)
    try:
        response = session.get("https://api.github.com/")
    except RetryError as err:
        print(f"Error: {err}")
```
